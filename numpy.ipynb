```python
import numpy as np

arr = np.array([10,20,30,40,50])
print(arr.sum())


# basics functions of numpy
print(np.sum(arr))
print(np.min(arr))
print(np.max(arr))
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))
print(np.var(arr)) # std * std
print(np.prod(arr))
print(np.cumsum(arr))
print(np.cumprod(arr))
print(np.diff(arr))
print(np.nonzero(arr))
print(np.sort(arr))
print(np.argsort(arr))
print(np.argmin(arr))
print(np.argmax(arr))
print(np.zeros([5,5])) # numpy array
print("\n\n")

print(np.identity(4))

print(np.eye(5))
```

```python
# matrix multiplicationa
import numpy as np

a = np.array([[1,2,3],[2,3,4]])
b = np.array([[1,2],[2,3],[3,4]])

print(np.dot(a,b))
print(a @ b)
print(np.matmul(a,b))
```

```python
# Transpose 

A = np.array([[1,2,3],[4,5,6]])
print(A.T)

print(np.transpose(A))
```

```python
# linear algebra

a = np.array([[1,2,-1],[2,1,1],[3,-1,2]])
b= np.array([3,7,4])
print(np.linalg.solve(a,b))
```

```python
#determinent of a matrix

d = np.array([[1,2],[3,4]])
print(np.linalg.det(d))
```

```python
#Inverse of a matrix

d = np.array([[1,2,3],[3,4,6], [7,8,9]])
print(np.linalg.inv(d))
```

```python
# Trace of a matrics =  sum of a diagnal of a square matrix


d = np.array([[1,2,3],[3,4,6], [7,2,9]])
print(np.trace(d))
```

```python
#linespace

values = np.linspace(1,10,num=10)
print(values)
```

```python
# empty function

d = np.empty((3,3), dtype =int)
print(d)
```

```python
# flattened array to create a 1d array

d = np.array([[1,2,3],[3,4,6], [7,2,9]])
flt_arr = d.flatten()

print(d)
print(flt_arr)
```

```python
# ravel similarly to flatten but more efficient memory
d = np.array([[1,2,3],[3,4,6], [7,2,9]])
rav_arr = d.ravel()

print(d)
print(rav_arr)
```

```python
# hstack() function
d = np.array([[1,2,3],[3,4,6], [7,2,9]])
b = np.array([[1,2,3],[3,4,6], [7,2,10]])

print(np.hstack((d,b)))

# vstack()
print(np.vstack((d,b)))
```
